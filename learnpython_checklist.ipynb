# 1. Catch-up

> If you didn't attend any of the workshops, you won't have anything installed on your computer. You have two options. 
> 1. **easy** Use Google Colab! For any of these files, you can open them in Google Colab. Go to Google Drive, [install Google Colab](https://colab.research.google.com/), and then open the file in Google Colab, for example by using the strategies outlined here: https://www.youtube.com/watch?v=-Z43Xx6X6wE  
> 2. **medium-difficult** Go through the [Membership Setup Slides](https://docs.google.com/presentation/d/1-8bAkQRxD2mr20zIvDxBqKLBW_HPPSaSttsOYmwQShE/edit) and the [README.md document](https://github.com/IHEDelftCodingClub/intro_to_python) to setup your personal computer with a python coding environment
> 3. **easy, but not personally recommended** Install Anaconda. Anaconda has all the installs bundled together, but that means it installs a lot of things you will never use, and it will likely cause you problems later on if you continue to grow as a developer.
